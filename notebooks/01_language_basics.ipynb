## 1. Lexical structure

* **Indentation** defines blocks (no braces). Use 4 spaces.
* **One line, one statement.** Continue a line inside `()`, `[]`, `{}` (preferred over `\`).
* **Identifiers**: start with a letter or `_`, then letters/digits/`_`. Case-sensitive.
* **Keywords** are reserved - list them with the `keyword` module.
* **Operators**: arithmetic `+ - * / // % **`, comparison `< <= > >= == !=`, logical `and or not`,
  identity `is`, membership `in`.
* **Delimiters**: `() [] {} , : ; .`

```python
import keyword

print(len(keyword.kwlist), "keywords")
print(keyword.kwlist)
```

```python
total = (1 + 2 +
         3 + 4)          # implicit line continuation inside parentheses
print(total)
```

## 2. Everything is an object, typing is dynamic

Names are *labels* attached to objects; the **type belongs to the object, not to the name**.

```python
a = 3
print(type(a), id(a))
a = "hello"                # same name, now a different object of a different type
print(type(a), id(a))
print(a + a, 2 * a)        # operators are overloaded per type
```

## 3. Numbers

```python
print(7 * 3., 2 ** 10, 8 % 3)
print("true division:", 3 / 2, "| floor division:", 3 // 2, "| divmod:", divmod(17, 5))
print("big ints are exact:", 2 ** 100)

# Python 3 vs Python 2: `/` is ALWAYS true division. The slides' `3/2 == 1` is Python 2 behaviour.
print(-7 // 2, -7 % 2)     # floor division rounds toward -infinity
```

## 5. Booleans, truthiness, `==` vs `is`

Falsy values: `0`, `0.0`, `0j`, `None`, `False`, and any **empty** container / string.
Everything else is truthy.

* `==` compares **values**
* `is` compares **identity** (same object). Use `is` only for `None` / `True` / `False`.

```python
for value in [0, 0.0, "", [], {}, set(), None, False, " ", [0], "False"]:
    print(f"{value!r:>8} -> {bool(value)}")
```

```python
x = [1, 2, 3]
y = [1, 2, 3]
print(x == y, x is y)
print(None is None, 5 in x, 2 in x)
```

### Short-circuiting
`and` stops at the first falsy operand; `or` stops at the first truthy one. Both return the
operand that decided, not necessarily a bool.

```python
def noisy(label, value):
    print("evaluating", label)
    return value

print(noisy("A", False) and noisy("B", True))   # B never evaluated
print(noisy("C", 0) or noisy("D", "fallback"))  # classic default idiom
```

## 6. Mutable vs immutable (the slides call this *very important*)

| Immutable | Mutable |
|---|---|
| `int, float, complex, bool, str, tuple, frozenset, bytes` | `list, dict, set, bytearray`, most user classes |

Assignment **never copies**; it binds another name to the same object.

```python
# Immutable: "changing" the value rebinds the name
a = 32
b = a
a = 10
print(a, b)

# Mutable: both names see the in-place change
l = [32, 10]
r = l
l[1] = 5
print(l, r, l is r)
```

```python
# Copies: shallow vs deep
import copy

A = [1, 2]
B = [A, 3]                 # B contains A
C = B.copy()               # shallow copy (same as B[:] or list(B))
D = copy.deepcopy(B)
A.append("changed")
print("B:", B)
print("C (shallow):", C)   # inner list is shared -> sees the change
print("D (deep)   :", D)   # fully independent
```

```python
# Tuples are immutable, but may *contain* mutable objects
t = ([1, 2], 3)
t[0].append(99)
print(t)
try:
    t[1] = 4
except TypeError as err:
    print("TypeError:", err)
```

## 7. Sequences: strings, lists, tuples

Shared operations: `+` (concat), `*` (repeat), `in`, `len`, indexing `s[i]`, slicing `s[i:j:k]`,
`min/max`, `.index`, `.count`, iteration.

```python
l = [1, 2, 3, 4, 5]
print(l[2], l[-1], l[-2], l[2:4], l[3:], l[:3], l[::2], l[::-1])
print(l[10:], "<- slicing never raises IndexError")

# Slicing returns a NEW list (shallow copy)
m = l[:]
m[0] = 99
print(l, m)

# Slice assignment
l[1:3] = ["a", "b", "c"]
print(l)
```

```python
# Slicing rules:  s[start:stop:step]  ->  start <= i < stop, so len == stop - start (step 1)
word = "abcdefgh"
print(word[2:5], len(word[2:5]), word[-3:], word[:-3], word[5:2:-1])
```

```python
# List methods
l = [1, 2, 3, 4, 5]
l.append(6)
l.extend([7, 8])
l.insert(0, 0)
print(l)
print("pop() ->", l.pop(), "| pop(0) ->", l.pop(0))
print(l)
l.remove(3)
l.reverse()
print(l, l.index(5), l.count(5))
l.sort(reverse=True)               # in-place, returns None
print(l, sorted(l), l)             # sorted() returns a new list
```

```python
# Tuples: immutable, hashable (if contents are), great as light records / return values
t = 12345, 54321, "hello!"
print(t, t[0], len(t), t.count(12345), t.index("hello!"))
single = (1,)                      # the comma makes the tuple, not the parentheses
print(type(single), type((1)))
```

## 8. Dictionaries

Hash tables mapping *hashable* keys to values. Since Python 3.7 they **preserve insertion order**
(the slides' "unordered" is an older statement). Lookup, insert, delete are O(1) on average.

```python
tel = {"emmanuelle": 5752, "sebastian": 5578}
tel["francis"] = 5915
print(tel, tel["sebastian"], "francis" in tel, len(tel))
print(list(tel.keys()), list(tel.values()))
print(tel.get("nobody"), tel.get("nobody", -1))
```

```python
for key, value in tel.items():
    print(f"{key:>10}: {value}")
```

```python
# More methods
d = {"a": 1}
d.update({"b": 2, "a": 10})
print(d)
print("setdefault ->", d.setdefault("c", 3), "| pop ->", d.pop("b"))
print(d)
print("popitem ->", d.popitem())      # removes the LAST inserted pair
print(d)
merged = {"x": 1} | {"y": 2}         # Python 3.9+
print(merged)
```

```python
# Keys must be hashable (immutable): tuples yes, lists no
grid = {(0, 0): "origin", (1, 2): "point"}
print(grid[(1, 2)])
try:
    {[1, 2]: "nope"}
except TypeError as err:
    print("TypeError:", err)
```

```python
# Handy patterns: counting & grouping
from collections import Counter, defaultdict

words = "the cat and the hat and the bat".split()
counts = {}
for w in words:
    counts[w] = counts.get(w, 0) + 1
print(counts, Counter(words).most_common(2))

groups = defaultdict(list)
for w in words:
    groups[len(w)].append(w)
print(dict(groups))
```

## 9. Sets and frozensets

Unordered collections of **unique, hashable** items; membership test is O(1).

```python
s = set(("a", "b", "c", "a"))
print(s, len(s))
A, B = {1, 2, 3, 4}, {3, 4, 5}
print("union", A | B, "| intersection", A & B, "| difference", A - B, "| symmetric", A ^ B)
print("subset?", {1, 2} <= A, "| disjoint?", A.isdisjoint({9}))

s.add("d"); s.discard("zzz")           # discard never raises; remove() does
print(s)

fs = frozenset([1, 2, 3])               # immutable & hashable -> can be a dict key / set member
print({fs: "ok"})

# Remove duplicates (order NOT guaranteed); keep order with dict.fromkeys
nums = [3, 1, 3, 2, 1]
print(list(set(nums)), list(dict.fromkeys(nums)))
```

## 10. Comprehensions

A concise way to build a container; reads like set-builder notation: `k = { x² | x ∈ {0,1,2,3} }`.

```python
k = [x ** 2 for x in range(4)]
print(k)
print([x for x in range(20) if x % 3 == 0])
print([(i, j) for i in range(3) for j in range(3) if i < j])
print({w: len(w) for w in ["cool", "powerful", "readable"]})      # dict comprehension
print({c for c in "mississippi"})                                  # set comprehension
matrix = [[1, 2, 3], [4, 5, 6]]
print([v for row in matrix for v in row], [list(col) for col in zip(*matrix)])
```

## 11. Packing, unpacking, walrus

* **Packing**: `coords = 1, 2, 3` (a tuple)
* **Unpacking**: `x, y, z = coords`, with star: `first, *rest = items`
* **Walrus** `:=` assigns inside an expression (3.8+).

```python
coords = (1, 2, 3)
x, y, z = coords
first, *middle, last = [10, 20, 30, 40, 50]
print(x, y, z, first, middle, last)

a, b = 1, 2
a, b = b, a                # swap without a temp variable
print(a, b)

for i, (name, score) in enumerate([("ann", 9), ("bob", 7)], start=1):
    print(i, name, score)
```

```python
# Walrus: compute once, use twice
data = [1, 5, 12, 7]
if (n := len(data)) > 3:
    print(f"long list with {n} items")

import re
if (m := re.search(r"\d+", "abc123def")):
    print("found", m.group())
```
