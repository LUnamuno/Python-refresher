# TennisTournament

[Codility task](https://app.codility.com/programmers/trainings/3/tennis_tournament/) - Exercise 3, 2017 Contest

`P` players are registered and `C` courts are reserved. A game needs two players and one court. `solution(P, C)` returns the maximum number of games that can be played **at the same time**.

| P | C | result |
|---|---|--------|
| 5 | 3 | 2 |
| 10 | 3 | 3 |

Assumption: `P` and `C` are in `[1..30,000]`.

## Idea

Two independent upper bounds:

- players: every game takes two of them, so at most `P // 2` games;
- courts: at most one game per court, so at most `C` games.

Both must hold, so the answer is `min(P // 2, C)`. Time and space O(1).

```python
def solution(P, C):
    return min(P // 2, C)
```

## Official examples

```python
assert solution(5, 3) == 2
assert solution(10, 3) == 3
print("examples OK")
```

## Edge cases

```python
assert solution(1, 1) == 0                      # a single player cannot play
assert solution(2, 1) == 1
assert solution(3, 1) == 1                      # the third player has no partner and no court
assert solution(2, 30_000) == 1                 # many courts, few players
assert solution(30_000, 1) == 1                 # many players, one court
assert solution(1, 30_000) == 0
assert solution(30_000, 30_000) == 15_000       # courts are not the limit
assert solution(29_999, 30_000) == 14_999       # odd number of players
assert solution(30_000, 15_000) == 15_000       # both limits coincide
print("edge cases OK")
```

## Exhaustive check

The reference does not use the formula: it looks for the largest `g` such that `g` games are feasible (`2 * g <= P` players and `g <= C` courts). All pairs with `P, C` in `[1..120]` are checked.

```python
def brute(P, C):
    best = 0
    for g in range(0, P + 1):
        if 2 * g <= P and g <= C:
            best = g
    return best


for P in range(1, 121):
    for C in range(1, 121):
        assert solution(P, C) == brute(P, C), (P, C)
print("matches the reference for all 14,400 pairs (P, C) in [1..120]")
```

## Notes

- Integer division `//` discards the player without a partner.
- `min` of independent capacity limits is a common pattern: the bottleneck decides.
- The statement says performance is not assessed; here it is constant time anyway.
