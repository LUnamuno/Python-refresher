# StrSymmetryPoint

[Codility task](https://app.codility.com/programmers/trainings/4/str_symmetry_point/) - Exercise 4, Algorithmic skills

`solution(S)` returns the index of a character such that the part of the string to its left is the **reversal** of the part to its right, or `-1` if no such index exists. Reversing the empty string gives the empty string.

| S | result |
|---|--------|
| `"racecar"` | 3 (`"rac"` and `"car"`) |
| `"x"` | 0 (both sides empty) |

Assumption: the length of `S` is in `[0..2,000,000]`.

## Idea

For the two sides to have the same length the character must be the **middle** one, which exists only when `len(S)` is odd. So there is at most one candidate: `len(S) // 2`. It is valid iff the whole string is a palindrome.

- even length (including the empty string): `-1`;
- odd length: `len(S) // 2` if `S == S[::-1]`, otherwise `-1`.

Time O(N), space O(N) for the reversed copy.

```python
def solution(S):
    n = len(S)
    if n % 2 == 0:
        return -1
    return n // 2 if S == S[::-1] else -1
```

## Official examples

```python
assert solution("racecar") == 3
assert solution("x") == 0
print("examples OK")
```

## Edge cases

```python
assert solution("") == -1                 # no character at all
assert solution("ab") == -1               # even length
assert solution("aa") == -1               # even palindrome: no centre character
assert solution("aba") == 1
assert solution("abc") == -1              # odd, but not a palindrome
assert solution("aaaaa") == 2
assert solution("abcba") == 2
assert solution("abcda") == -1
print("edge cases OK")
```

## Exhaustive brute force

The reference literally follows the statement: try every index and compare the left part with the reversed right part. All strings over `{a, b}` up to length 10 are checked (2047 strings).

```python
import itertools


def brute(S):
    for i in range(len(S)):
        if S[:i] == S[i + 1:][::-1]:
            return i
    return -1


checked = 0
for n in range(0, 11):
    for letters in itertools.product("ab", repeat=n):
        s = "".join(letters)
        assert solution(s) == brute(s), s
        checked += 1
print(f"matches brute force on all {checked} strings over {{a, b}} up to length 10")
```

## Performance: strings of up to 2,000,000 characters

```python
import time

N = 2_000_000
half = "ab" * 499_999 + "a"                      # 999,999 characters
palindrome = half + "c" + half[::-1]             # 1,999,999 characters
broken = half + "c" + "x" + half[::-1][1:]       # same length, not a palindrome
even = "a" * N                                   # 2,000,000 characters

for name, s, expected in [
    ("odd palindrome", palindrome, len(half)),
    ("odd, not a palindrome", broken, -1),
    ("even length", even, -1),
]:
    start = time.perf_counter()
    result = solution(s)
    print(f"{name:24s} len={len(s):>9} -> {result:>8} in {time.perf_counter() - start:.4f}s")
    assert result == expected
```

## Notes

- "Left is the reversal of right" for a single index is exactly the definition of an odd palindrome centred there.
- `S[::-1]` makes a copy; a two-pointer loop avoids the extra memory but is slower in pure Python.
- Trying every index and slicing is O(N^2): that is what the 2,000,000 limit is designed to reject.
