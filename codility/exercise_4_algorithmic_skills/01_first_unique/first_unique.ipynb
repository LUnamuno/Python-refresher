# FirstUnique

[Codility task](https://app.codility.com/programmers/trainings/4/first_unique/) - Exercise 4, Algorithmic skills

A *unique number* occurs exactly once in `A`. `solution(A)` returns the unique number that appears **first** in the array (the value, not the index), or `-1` if there is none.

| A | result |
|---|--------|
| `[4, 10, 5, 4, 2, 10]` | 5 |
| `[1, 4, 3, 3, 1, 2]` | 4 |
| `[6, 4, 4, 6]` | -1 |

Assumptions: `N` in `[1..100,000]`, values in `[0..10^9]`.

## Idea

1. Count the occurrences of every value in one pass (`collections.Counter`).
2. Scan the array again in order and return the first value whose count is 1.

`A.count(x)` inside a loop would be O(N^2). Here: O(N) time, O(N) space.

```python
from collections import Counter


def solution(A):
    counts = Counter(A)
    for value in A:
        if counts[value] == 1:
            return value
    return -1
```

## Official examples

```python
assert solution([4, 10, 5, 4, 2, 10]) == 5
assert solution([1, 4, 3, 3, 1, 2]) == 4
assert solution([6, 4, 4, 6]) == -1
print("examples OK")
```

## Edge cases

```python
assert solution([7]) == 7                         # a single element is unique
assert solution([0]) == 0                         # value 0 must not be confused with "not found"
assert solution([0, 0, 1]) == 1
assert solution([5, 5, 5]) == -1                  # an odd number of copies is still not unique
assert solution([1, 2, 3]) == 1                   # all unique: the first one
assert solution([2, 2, 3, 3, 4]) == 4             # unique at the very end
assert solution([10**9, 10**9, 0]) == 0
print("edge cases OK")
```

## Brute-force check

The reference uses `list.count`, which is O(N^2) but obviously correct.

```python
import random


def brute(A):
    for value in A:
        if A.count(value) == 1:
            return value
    return -1


rng = random.Random(1)
for _ in range(5000):
    n = rng.randint(1, 15)
    A = [rng.randint(0, 6) for _ in range(n)]
    assert solution(A) == brute(A), A
print("matches brute force on 5000 random arrays")
```

## Performance

```python
import time

N = 100_000
cases = {
    "all distinct": list(range(N)),
    "all equal": [3] * N,
    "unique at the end": [1] * (N - 1) + [2],
    "random, wide values": [rng.randint(0, 10**9) for _ in range(N)],
    "random, many ties": [rng.randint(0, 1000) for _ in range(N)],
}
for name, A in cases.items():
    start = time.perf_counter()
    result = solution(A)
    print(f"{name:22s} -> {result:>10} in {time.perf_counter() - start:.4f}s")
```

## Notes

- Return the **value**; the position only decides which unique value wins.
- Python's `dict`/`Counter` is a hash table: O(1) average per operation.
- Similar problems: first non-repeating character, "single number" (with XOR when every other value appears twice).
