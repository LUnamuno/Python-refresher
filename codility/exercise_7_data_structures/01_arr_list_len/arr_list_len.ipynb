# ArrListLen

[Codility task](https://app.codility.com/programmers/trainings/7/arr_list_len/) - Exercise 7, Data Structures - *easy*

Array `A` encodes a singly linked list:

- the head is at index `0`;
- the node at index `K` has value `A[K]`;
- value `-1` marks the last node, otherwise the successor of node `K` is at index `A[K]`.

`solution(A)` returns the number of nodes of the list.

For `A = [1, 4, -1, 3, 2]` the nodes are at indices `0 -> 1 -> 4 -> 2`, so the answer is `4`.

Assumptions: `N` in `[1..200,000]`, each element in `[-1..N-1]`, and the list is always finite.

## Idea

Just follow the pointers: count the current node, then jump to `A[index]` until the value is `-1`. The node holding `-1` **is** counted.

Time O(N), extra space O(1). The statement guarantees there is no cycle, so no visited set is needed.

```python
def solution(A):
    length = 0
    index = 0
    while index != -1:
        length += 1
        index = A[index]
    return length
```

## Official example

```python
A = [1, 4, -1, 3, 2]
assert solution(A) == 4
print("example OK")
```

## Edge cases

```python
assert solution([-1]) == 1                      # a single node
assert solution([1, -1]) == 2
assert solution([-1, 0, 1, 2]) == 1             # the head is already the last node
assert solution([1, 2, 3, 4, -1]) == 5          # chain in index order
assert solution([4, 0, 1, 2, -1]) == 2          # 0 -> 4 -> -1 (indices 1, 2, 3 are not reachable)
print("edge cases OK")
```

## Randomised check against a constructed list

The oracle goes the other way round: **build** an array that encodes a list of known length `L` (choose `L` distinct indices starting at `0`, link them in order, end with `-1`), fill the unreachable positions with arbitrary valid values (noise) and check that `solution` recovers `L`.

```python
import random


def build_case(n, length, rng):
    """Array of n elements whose list has exactly `length` nodes."""
    path = [0] + rng.sample(range(1, n), length - 1)
    A = [rng.randint(-1, n - 1) for _ in range(n)]          # noise everywhere
    for here, nxt in zip(path, path[1:]):
        A[here] = nxt
    A[path[-1]] = -1
    return A


rng = random.Random(7)
for _ in range(2000):
    n = rng.randint(1, 30)
    length = rng.randint(1, n)
    A = build_case(n, length, rng)
    assert solution(A) == length, (A, length)
print("randomised check OK")
```

## Large input

```python
import time

rng = random.Random(1)
N = 200_000
A = build_case(N, N, rng)                       # a list visiting every index, shuffled
start = time.perf_counter()
assert solution(A) == N
print(f"N = {N}: OK in {time.perf_counter() - start:.3f}s")
```

## Notes

- Pitfall: returning the number of *jumps* (length - 1) instead of the number of *nodes*.
- The task says performance is not assessed, but the simple loop is already linear.
- If cycles were possible you would add a `visited` set (or Floyd's tortoise and hare) and decide what to return.
