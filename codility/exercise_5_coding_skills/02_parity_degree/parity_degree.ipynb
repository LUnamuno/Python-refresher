# ParityDegree

[Codility task](https://app.codility.com/programmers/trainings/5/parity_degree/) - Exercise 5, Coding skills - *easy*

Given a positive integer `N`, find the **maximum `K` such that `N mod 2^K == 0`**, i.e. the exponent of the highest power of 2 that divides `N`.

Example: `N = 24` -> `3`, because `2^3 = 8` divides 24 and `2^4 = 16` does not.

Assumption: `N` in `[1..1,000,000,000]`.

## Ideas

1. **Loop:** divide by 2 while the number is even and count the divisions. At most about 30 iterations for `N <= 10^9`.
2. **Bit trick:** `K` is the number of trailing zero bits of `N`. `N & -N` keeps only the lowest set bit (which is `2^K`), and `bit_length() - 1` gives `K`.

For `24 = 0b11000`: the lowest set bit is `0b1000 = 8`, whose bit length is 4, so `K = 3`.

```python
def solution(N):
    """Bit trick: number of trailing zeros of N."""
    return (N & -N).bit_length() - 1


def solution_loop(N):
    """Plain loop: how many times can we halve N?"""
    k = 0
    while N % 2 == 0:
        N //= 2
        k += 1
    return k
```

## Official example

```python
for f in (solution, solution_loop):
    assert f(24) == 3
print("example OK")
```

## Edge cases

```python
for f in (solution, solution_loop):
    assert f(1) == 0                              # 2^0 divides 1
    assert f(2) == 1
    assert f(3) == 0                              # odd numbers -> 0
    assert f(12) == 2
    assert f(2**29) == 29                         # 536,870,912: largest power of two <= 10^9
    assert f(10**9) == 9                          # 10^9 = 2^9 * 5^9
    assert f(999_999_999) == 0                    # largest odd number in range
    for k in range(30):                           # every power of two, and powers times an odd factor
        assert f(2**k) == k
        if 3 * 2**k <= 10**9:
            assert f(3 * 2**k) == k
print("edge cases OK")
```

## Check against the definition

The reference literally implements the statement: the **maximum** `K` with `N % 2**K == 0`, trying every `K` from 0 to 30.

```python
def by_definition(N):
    return max(k for k in range(31) if N % (2**k) == 0)


for n in range(1, 200_001):
    expected = by_definition(n)
    assert solution(n) == expected, n
    assert solution_loop(n) == expected, n

import random
rng = random.Random(0)
for _ in range(20_000):
    n = rng.randint(1, 10**9)
    assert solution(n) == solution_loop(n) == by_definition(n), n
print("both solutions match the definition for N = 1..200000 and 20000 random values")
```

## Notes

- `N & -N` works because `-N` is `~N + 1`: all bits above the lowest set bit are flipped, so only that bit survives the `&`.
- `int.bit_length()` is available on every Python 3. Do not use `bit_count()` here: it counts the set bits, not the position of the lowest one.
- `N` is positive by assumption; `0` would make the loop run forever and the trick return `-1`.
